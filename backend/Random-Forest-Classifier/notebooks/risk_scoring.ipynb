```python
def calculate_risk(approve_probability):

    risk_score = (1 - approve_probability) * 100

    if risk_score <= 30:
        risk_level = "LOW"
    elif risk_score <= 60:
        risk_level = "MEDIUM"
    else:
        risk_level = "HIGH"

    if approve_probability >= 0.50:
        decision = "APPROVE"
    else:
        decision = "REJECT"

    return {
        "approval_probability": round(approve_probability, 3),
        "risk_score": round(risk_score, 2),
        "risk_level": risk_level,
        "decision": decision
    }
```

```python
print(calculate_risk(0.90))
```

```python
print(calculate_risk(0.62))
```

```python
def get_risk_factors(customer):
    
    factors = []

    if customer["Credit_Score"] < 650:
        factors.append("Low credit score")

    if customer["Debt_to_Income_Ratio"] > 40:
        factors.append("High debt-to-income ratio")

    if customer["Credit_Utilization"] > 70:
        factors.append("High credit utilization")

    if customer["Previous_Defaults"] > 0:
        factors.append("Previous loan default")

    if customer["Missed_Payments"] > 2:
        factors.append("Multiple missed payments")

    if customer["Maximum_Days_Past_Due"] > 30:
        factors.append("High payment delinquency")

    if customer["Recent_Credit_Enquiries"] > 3:
        factors.append("Multiple recent credit enquiries")

    return factors
```

```python
def generate_credit_assessment(customer, approve_probability):

    result = calculate_risk(
        approve_probability
    )

    risk_factors = get_risk_factors(
        customer
    )

    result["risk_factors"] = risk_factors

    return result
```

```python
customer = {
    "Credit_Score": 620,
    "Debt_to_Income_Ratio": 48,
    "Credit_Utilization": 75,
    "Previous_Defaults": 1,
    "Missed_Payments": 3,
    "Maximum_Days_Past_Due": 45,
    "Recent_Credit_Enquiries": 4
}
```

```python
result = generate_credit_assessment(
    customer,
    0.42
)

print(result)
```
